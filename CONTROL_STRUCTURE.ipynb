**IF statement**

```python
a = 12
b = 100
if b  > a:
    print("b is greater than a.")
```

ELIF statements

```python
a = 12
b = 12
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
```

ELSE statements

```python
a = 200
b = 33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
else:
    print("a is greater than b")
```

Short Hand IF statement

```python
if a > b: print("a is greater than b")
```

Short Hand IF...ELSE statement

```python
a = 2
b = 330
print("A") if a > b else print("B")
```

AND-BOTH conditions are true

```python
a = 200
b = 33
c = 500
if a > b and c > a:
    print("Both conditions are True")
```

OR-AT least one of the condition is true

```python
a = 200
b - 33
c = 500
if a > b or a > c:
    print("At least one of the conditions is True")
```

NESTED IF statement

```python
# Create a python program that wil display above 10, above 20, above 30, above 40, but below 50
# +5 Points ╰(*°▽°*)╯
x = 41
print("Above 10") if x > 10 else print("Below 10")
print("Above 20") if x > 20 else print("Below 20")
print("Above 30") if x > 30 else print("Below 30")
print("Above 40") if x > 40 else print("Below 40")
print("Below 50") if x < 50 else print("Above 50")
```

Example 3 : Write a program to determine if the grades are:
    if:
        grade >= 75, "Passed"
        grade = 74, "Remedial"
        grade < 74, "Failed"
        Note: if grade is a negative number, it will not be accepted


```python
# +5
grade = int(input("Input Grade: "))
if grade >= 75: print("Passed!")
elif grade == 74: print("Remedial!")
elif grade > -1: print("Failed!")
else: print("Invalid Input!")
```
