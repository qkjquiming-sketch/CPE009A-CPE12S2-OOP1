### Distance Calculator

This program calculates the distance between two points $(x_1, y_1)$ and $(x_2, y_2)$ using the formula:

$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$

```python
print("--- DISTANCE CALCULATOR ---")

x1 = float(input("Enter x-coordinate for Point 1 (x1): "))
y1 = float(input("Enter y-coordinate for Point 1 (y1): "))

x2 = float(input("Enter x-coordinate for Point 2 (x2): "))
y2 = float(input("Enter y-coordinate for Point 2 (y2): "))

distance = ((x2 - x1)**2 + (y2 - y1)**2)**0.5

print("--- RESULTS ---")
print("%.9f" % distance)
```
